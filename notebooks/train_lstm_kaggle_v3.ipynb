```python
from pathlib import Path
import runpy
INPUT_ROOT = Path('/kaggle/input') if Path('/kaggle/input').exists() else Path('/content/stock-agent-v3')
scripts = list(INPUT_ROOT.rglob('train_lstm_kaggle_v3.py'))
if len(scripts) != 1:
    raise FileNotFoundError('Attach exactly one train_lstm_kaggle_v3.py as input; alternatively paste that .py code into a cell.')
training = runpy.run_path(str(scripts[0]), run_name='v3_training_import')
datasets = list(INPUT_ROOT.rglob('vci_ohlcv_lstm_v3.csv'))
if len(datasets) != 1:
    raise FileNotFoundError('Attach exactly one NEW V3 CSV with its manifest/protocol/runtime ZIP in the same folder.')
DATA_ARGS = ['--data', str(datasets[0])]
OUTPUT_ROOT = Path('/kaggle/working') if Path('/kaggle/working').exists() else Path('/content')
print('Training script loaded; no training has started.')
```
